# Notebook 3 – Ghép nhiều ảnh thành panorama

Ảnh dài `images/img4.jpg` được cắt thành 8 ảnh con (rộng 600 px, lệch nhau 250 px, xoay ngẫu nhiên −10°…10°) để giả lập 8 bức chụp liên tiếp. Sau đó gọi đúng quy trình mà giao diện dùng: `Panorama.get_panorama()`.

Thuật toán ghép **bắt đầu từ ảnh giữa** (ảnh tham chiếu), ảnh bên trái nhân dồn H, ảnh bên phải nhân dồn H⁻¹.

```python
import sys, os
sys.path.insert(0, os.path.abspath("python"))   # notebook đặt ở thư mục gốc repo
import numpy as np
import matplotlib.pyplot as plt
import image_ops as ops
from sift_extractor import SIFT
from panorama_stitching import Panorama
from run_panorama import read_rgb              # chỉ dùng OpenCV để ĐỌC file ảnh
from general_functions import show_images, normalize_img
%matplotlib inline
```

```python
img = read_rgb("images/img4.jpg") / 255.0
rng = np.random.default_rng(0)
img_splits = [ops.rotate(img[:, 250 * i: 600 + 250 * i], rng.integers(-10, 10)) for i in range(8)]
show_images(img_splits, [f"ảnh {i + 1}" for i in range(8)], figsize=(18, 3), show_axis=False)
```

```python
pano = Panorama(verbose=True)
result = pano.get_panorama(img_splits)
for i, m in enumerate(pano.match_info):
    print(f"cặp {i + 1}-{i + 2}: {m['matches']} cặp khớp, {m['inliers']} inlier")
```

```python
show_images([img, result], ["Ảnh gốc", "Ảnh ghép được"], figsize=(18, 5), show_axis=False)
```
